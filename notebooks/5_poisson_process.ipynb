# Poisson Process

```python
import numpy as np
from matplotlib import pyplot as plt

T = 1
L = 100

dt = T/L

lam = 5

N = np.zeros(L+1)

for i in range(1, L+1):
    p = np.exp(-lam*dt)
    F = p
    x = 0.

    u = np.random.rand()

    while u > F:
        x += 1
        p = p*lam*dt/x
        F += p

    N[i] += N[i-1] + x

t = np.linspace(0, T, L+1)

plt.plot(t, N, 'o')
plt.xlabel('t')
plt.ylabel('N')
```
